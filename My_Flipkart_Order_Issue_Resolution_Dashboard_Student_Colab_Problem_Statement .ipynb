## Section 7: Data Cleaning and Standardization

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

```python
# SECTION 7 — Cleaning and standardization
# BROKEN: This code mutates only one column and ignores dates, amounts, and other files.

def clean_text(s):
    """Spaces/underscore hatao aur har label ki sabse common spelling use karo."""
    s = s.str.strip().str.replace("_", " ")
    best = s.groupby(s.str.lower()).agg(lambda x: x.value_counts().index[0])
    return s.str.lower().map(best)

def clean_date(s):
    d = pd.to_datetime(s, errors="coerce", format="%Y-%m-%d")
    swapped = pd.to_datetime(s, errors="coerce", format="%Y-%d-%m")
    return d.fillna(swapped)

orders_clean = all_data["orders"].copy()
for col in ["order_status", "payment_method", "delivery_city", "delay_reason", "seller_category"]:
    orders_clean[col] = clean_text(orders_clean[col])
for col in ["order_date", "promised_delivery_date", "actual_delivery_date"]:
    orders_clean[col] = clean_date(orders_clean[col])
orders_clean["order_amount"] = pd.to_numeric(orders_clean["order_amount"], errors="coerce")
orders_clean["order_amount_valid"] = orders_clean["order_amount"].where(orders_clean["order_amount"] > 0)

customers_clean = all_data["customers"].copy()
customers_clean["customer_segment"] = clean_text(customers_clean["customer_segment"])
sellers_clean = all_data["sellers"].copy()

refunds_clean = all_data["refunds"].copy()
for col in ["refund_status", "refund_reason"]:
    refunds_clean[col] = clean_text(refunds_clean[col])
for col in ["refund_initiated_date", "refund_completed_date"]:
    refunds_clean[col] = clean_date(refunds_clean[col])

complaints_clean = all_data["complaints"].copy()
for col in ["complaint_type", "complaint_status", "sentiment_tag"]:
    complaints_clean[col] = clean_text(complaints_clean[col])
complaints_clean["complaint_date"] = clean_date(complaints_clean["complaint_date"])

tickets_clean = all_data["tickets"].copy()
for col in ["ticket_status", "assigned_team", "escalation_flag"]:
    tickets_clean[col] = clean_text(tickets_clean[col])
tickets_clean["ticket_created_date"] = clean_date(tickets_clean["ticket_created_date"])
tickets_clean["resolution_time_hours"] = pd.to_numeric(tickets_clean["resolution_time_hours"], errors="coerce")


REFERENCE_DATE = max(orders_clean["actual_delivery_date"].max(), refunds_clean["refund_completed_date"].max(),
                     complaints_clean["complaint_date"].max(), tickets_clean["ticket_created_date"].max())
print("REFERENCE_DATE:", REFERENCE_DATE.date())

cols = ["order_status", "payment_method", "delivery_city"]
cleaning_summary = pd.DataFrame({"column": cols,
                                 "unique_before": [all_data["orders"][col].nunique() for col in cols],
                                 "unique_after": [orders_clean[col].nunique() for col in cols]})
display(cleaning_summary)
display(orders_clean.head())

log_fix("Section 7", "Only order_status lowercased; dates, amounts, other files ignored", "clean_text + clean_date for all datasets", cleaning_summary.to_dict("records"))

# TODO:
# - trim whitespace
# - standardize casing and underscores
# - parse date columns safely
# - clean amount fields
# - standardize refund/complaint/ticket status fields
display(orders_clean.head())
```

```python
# SECTION 13 — Final feature table
# BROKEN: Raw merges can accidentally multiply rows.


refund_by_order = (refunds_df.groupby("order_id")
    .agg(refund_count=("refund_id", "count"), refund_failed=("is_failed", "sum"), refund_open_breach=("open_breach", "sum"),
         refund_exceeds=("refund_exceeds_order", "sum"), refund_max_age_days=("refund_age_days", "max"),
         refund_total=("refund_amount", "sum"), refund_priority_count=("refund_priority_candidate", "sum"))
    .reset_index())

final_feature_table = (orders_clean
    .merge(customers_clean[["customer_id", "customer_name", "customer_segment"]], on="customer_id", how="left")
    .merge(sellers_clean[["seller_id", "seller_name", "seller_rating"]], on="seller_id", how="left")
    .merge(refund_by_order, on="order_id", how="left")
    .merge(complaints_order, on="order_id", how="left")
    .merge(tickets_order, on="order_id", how="left"))


agg_cols = [col for table in (refund_by_order, complaints_order, tickets_order) for col in table.columns if col != "order_id"]
final_feature_table[agg_cols] = final_feature_table[agg_cols].fillna(0)
final_feature_table["customer_segment"] = final_feature_table["customer_segment"].fillna("Unknown")
final_feature_table["customer_name"] = final_feature_table["customer_name"].fillna("Customer")

merge_validation = pd.DataFrame([{"orders_rows": len(orders_clean), "final_rows": len(final_feature_table),
                                  "row_count_ok": len(orders_clean) == len(final_feature_table)}])
display(merge_validation)
display(final_feature_table[["order_id", "customer_segment", "delay_bucket", "refund_count", "complaint_count", "ticket_count"]].head())

log_fix("Section 13", "Raw merges multiply rows", "Aggregated refunds/complaints/tickets first, then merged", merge_validation.to_dict("records"))

# TODO:
# - aggregate refunds, complaints, and tickets before merging
# - merge customer and seller master data
# - prevent accidental row explosion
```

## Section 14: Priority Classification

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

**Datasets to use:**
- `final_feature_table`

**How to approach:**
- Create a scoring function.
- Combine delay, refund, complaint, ticket, and customer segment signals.
- Return priority level and reasons.

**Expected output format:**
- priority_summary table.
- final_feature_table with issue_score, issue_priority, priority_reasons.

**Hint:**
- Priority should not be based on delay alone.

```python
# SECTION 14 — Priority classification
# BROKEN: Priority logic is too shallow and ignores PRD rules.

def score_case(row):

    rules = [
        (3, row["delay_bucket"] == "Lost", "Shipment lost"),
        (3, row["delay_bucket"] == "Pending - Breached", "Order stuck past promised date"),
        (2, row["delay_bucket"] == "Severe", "Severe delivery delay"),
        (1, row["delay_bucket"] == "Moderate", "Moderate delivery delay"),
        (3, row["refund_failed"] > 0, "Refund failed"),
        (3, row["refund_open_breach"] > 0, "Open refund past SLA"),
        (2, row["refund_exceeds"] > 0, "Refund exceeds order amount"),
        (2, row["angry_complaints"] > 0, "Angry customer"),
        (1, row["angry_complaints"] == 0 and row["negative_complaints"] > 0, "Negative complaint"),
        (2, row["complaint_count"] > 1, "Repeated complaints"),
        (1, row["open_complaints"] > 0, "Complaint still open"),
        (2, row["escalated_tickets"] > 0, "Escalated ticket"),
        (1, row["ticket_sla_breaches"] > 0, "Ticket SLA breached"),
        (1, row["customer_segment"] in ("Premium", "Plus", "Business"), "Priority customer segment"),
    ]
    score = sum(points for points, cond, _ in rules if cond)
    reasons = "; ".join(reason for _, cond, reason in rules if cond) or "No active issue"
    priority = "High" if score >= 8 else "Medium" if score >= 4 else "Low"
    return pd.Series([score, priority, reasons], index=["issue_score", "issue_priority", "priority_reasons"])

final_feature_table[["issue_score", "issue_priority", "priority_reasons"]] = final_feature_table.apply(score_case, axis=1)

priority_summary = (final_feature_table.groupby("issue_priority")
    .agg(cases=("order_id", "size"), avg_score=("issue_score", "mean")).round(1)
    .reindex(["High", "Medium", "Low"]))
display(priority_summary)
display(final_feature_table.sort_values("issue_score", ascending=False)[["order_id", "issue_score", "issue_priority", "priority_reasons"]].head())

log_fix("Section 14", "Priority used delay only", "Score from delay, refund, complaint, ticket, customer segment + reasons", priority_summary["cases"].to_dict())

# TODO:
# - use delay severity
# - refund status and refund age
# - complaints and sentiment
# - ticket escalation and SLA
# - customer impact
display(final_feature_table[["order_id", "issue_priority"]].head())
```

**Developer handover note:** Complete this section as the Python Developer responsible for repairing the inherited internal tool.

```python
# SECTION 15 — Recommended action generator
# BROKEN: Recommendations are generic and not tied to business signals.

def recommend_action(row):
    if row["escalated_tickets"] > 0: return "Escalation Desk: call customer back using merged complaint + ticket view"
    if row["refund_failed"] > 0: return "Refund Ops: re-initiate failed refund and confirm with customer"
    if row["refund_open_breach"] > 0: return "Refund Ops: release pending refund (SLA breached)"
    if row["refund_exceeds"] > 0: return "Refund Ops: reconcile refund amount with order amount"
    if row["angry_complaints"] > 0 or row["complaint_count"] > 1: return "Support Ops: senior agent callback and resolution timeline"
    if row["delay_bucket"] == "Lost": return "Logistics Ops: raise lost-shipment claim, offer reship or refund"
    if row["delay_bucket"] == "Pending - Breached":
        team = "Seller Ops" if "Seller" in str(row["delay_reason"]) else "City Ops"
        return f"{team}: dispatch or cancel the stuck order ({row['delay_reason']})"
    if row["delay_bucket"] in ("Severe", "Moderate"): return f"City Ops: review courier delays in {row['delivery_city']}"
    return "None: no action needed, monitor"

final_feature_table["recommended_action"] = final_feature_table.apply(recommend_action, axis=1)
final_feature_table["action_owner"] = final_feature_table["recommended_action"].str.split(":").str[0]

action_summary = pd.crosstab(final_feature_table["action_owner"], final_feature_table["issue_priority"])
display(action_summary)
display(final_feature_table.sort_values("issue_score", ascending=False)[["order_id", "issue_priority", "recommended_action"]].head())

log_fix("Section 15", "Generic 'Review order' action", "Team-specific actions from business signals", final_feature_table["action_owner"].value_counts().to_dict())

# TODO:
# - create specific actions for refund operations, support, seller ops, and city ops
display(final_feature_table[["order_id", "issue_priority", "recommended_action"]].head())
```
